```python
order_total = float(input("Please enter the order total with no special characters: "))
day_of_week = input("Please enter the Day of the week: ")
if order_total < 20 and day_of_week in ["Monday", "Tuesday", "Wednesday", "Thursday"]:
    print("Your order total is less than $20 and it is a weekday. You will not be charged a delivery fee.")
if order_total == 20 and day_of_week in ["Monday", "Tuesday", "Wednesday", "Thursday"]:
    print("Your order total is exactly $20 and it is a weekday. You will not be charged a delivery fee.")
elif order_total < 20 and day_of_week in ["Friday", "Saturday", "Sunday"]:
    print("Your order total is less than $20 and it is a weekend. You will be charged a $5 delivery fee.")
elif order_total == 20 and day_of_week in ["Friday", "Saturday", "Sunday"]:
    print("Your order total is exactly $20 and it is a weekend. You will be charged a $5 delivery fee.")
elif order_total > 20 and day_of_week in ["Monday", "Tuesday", "Wednesday", "Thursday"]:
    print("Your order total is greater than $20 and it is a weekday. You will  be charged a $2 delivery fee.")
elif order_total > 20 and day_of_week in ["Friday", "Saturday", "Sunday"]:
    print("Your order total is greater than $20 and it is a weekend. You will be charged a $7 delivery fee.")
elif order_total < 0:
    print(f"You have entered {order_total}. Order total cannot be negative.")
elif order_total > 250:
    print(f"You have entered {order_total}. Order total cannot be greater than $250.")
else:
    print(f"You have entered {day_of_week}. Invalid input. Please Choose from Monday, Tuesday, Wednesday, Thursday, Friday, Saturday, or Sunday.")

```
